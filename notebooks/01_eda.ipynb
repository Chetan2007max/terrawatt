```python
import sys
sys.path.append('..')  # so we can import from src/

import pandas as pd
import matplotlib.pyplot as plt
from src.data_loader import load_raw_data
from src.hierarchy import REGION_MEMBERS, REGIONS

df = load_raw_data()
df = df.sort_values('yyyymmdd').reset_index(drop=True)
df.head()
```

```python
plt.figure(figsize=(16, 6))
for region in REGIONS:
    plt.plot(df['yyyymmdd'], df[f'{region}: EnergyMet'], label=region, alpha=0.7)
plt.plot(df['yyyymmdd'], df['India: EnergyMet'], label='National', color='black', linewidth=2)
plt.legend()
plt.title('Daily EnergyMet by Region (2013–2026)')
plt.xlabel('Date')
plt.ylabel('EnergyMet (MU)')
plt.show()
```

```python
one_year = df[(df['yyyymmdd'] >= '2023-01-01') & (df['yyyymmdd'] <= '2023-12-31')]

plt.figure(figsize=(16, 5))
plt.plot(one_year['yyyymmdd'], one_year['NR: EnergyMet'])
plt.title('Northern Region — Daily EnergyMet, 2023')
plt.xlabel('Date')
plt.ylabel('EnergyMet (MU)')
plt.show()
```

```python
plt.figure(figsize=(16, 5))
for region in REGIONS:
    one_year_r = one_year.copy()
    plt.plot(one_year_r['yyyymmdd'], one_year_r[f'{region}: EnergyMet'], label=region)
plt.legend()
plt.title('All Regions — Daily EnergyMet, 2023')
plt.xlabel('Date')
plt.ylabel('EnergyMet (MU)')
plt.show()
```

```python
one_month = df[(df['yyyymmdd'] >= '2023-06-01') & (df['yyyymmdd'] <= '2023-06-30')]
plt.figure(figsize=(14, 4))
plt.plot(one_month['yyyymmdd'], one_month['NR: EnergyMet'], marker='o')
plt.title('Northern Region — June 2023 (checking for weekly pattern)')
plt.xticks(rotation=45)
plt.show()
```
